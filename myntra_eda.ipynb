### Variables Description

The dataset consists of 7 variables representing Myntra product information.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Data Wrangling Code

# Check for duplicate rows to ensure data uniqueness
print("Duplicate rows:", df.duplicated().sum())

# Check for missing/null values across all columns
print("\nMissing values:\n", df.isnull().sum())

# Verify data types of each column for correctness
print("\nData Types:\n", df.dtypes)

# Create a new feature 'discount_amount' to represent actual price reduction
df['discount_amount'] = df['MRP'] - df['price']

# Categorize products into price ranges for better analysis
# Helps in segmenting products into Low, Medium, High, and Premium groups
df['price_range'] = pd.cut(df['price'],
                          bins=[0,500,1000,2000,5000],
                          labels=['Low','Medium','High','Premium'])

# Display final dataset shape after preprocessing
# Preview first few rows to confirm transformations
print("\nFinal Shape:", df.shape)
df.head()
````

### What all manipulations have you done and insights you found?

During data wrangling, the dataset was first validated by checking for duplicate entries and missing values, and it was confirmed that the data is clean with no duplicates or null values. The data types of all variables were also verified, ensuring that numerical columns such as price, MRP, discount_percent, ratings, and number_of_ratings are correctly formatted for analysis.

Feature engineering was performed to enhance the dataset by creating a new variable discount_amount, which captures the actual reduction in price from MRP, and price_range, which segments products into Low, Medium, High, and Premium categories. These transformations help in better grouping and comparison during analysis.

Initial insights reveal significant variation in product pricing and discount levels across different brands. The distribution of ratings and number of ratings indicates varying levels of customer engagement, suggesting that some products are significantly more popular than others. Overall, the dataset is well-structured and suitable for extracting meaningful insights through visualization

##### 1. Why did you pick the specific chart?

I used a histogram with KDE to understand the distribution of product prices. This chart helps in identifying how prices are spread, detecting skewness, and observing whether most products fall into a specific price range.

##### 2. What is/are the insight(s) found from the chart?

The chart shows a heavily right-skewed distribution, where most products are concentrated in the lower price range (roughly below ₹3000). There are very few products in the higher price range, but some extreme values (outliers) extend up to very high prices.

This indicates that the majority of products on Myntra are affordable, with only a small number of premium or luxury items.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

The concentration of products in the lower price range suggests that Myntra is targeting a mass-market audience, which can lead to higher sales volume and customer reach, positively impacting revenue.

This could highlight an opportunity to either strengthen the premium segment or optimize pricing strategies to better balance demand across price ranges.

#### Chart - 7

````python
# Chart - 7: Average Price vs Discount Percentage

# Goal: Identify trend between discount levels and pricing

# Create bins for discount %
df['discount_bin'] = pd.cut(df['discount_percent'], bins=10)

# Calculate average price for each bin
discount_trend = df.groupby('discount_bin')['price'].mean()

# Plot line chart
plt.figure(figsize=(10,5))

plt.plot(discount_trend.values, marker='o')

plt.title("Average Price across Discount Levels")
plt.xlabel("Discount Bins")
plt.ylabel("Average Price")

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
````

##### 1. Why did you pick the specific chart?

I chose a line plot because it effectively shows trends and patterns between discount levels and average price.

##### 2. What is/are the insight(s) found from the chart?

The relationship between discount and price is not linear. Very low discount levels show high average prices, while moderate discounts show more stable pricing. There is fluctuation across discount bins, indicating inconsistent pricing strategies.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

The inconsistent trend suggests that discounts are not uniformly applied across price segments. Optimizing discount strategies based on product pricing could improve sales efficiency.

Irregular pricing-discount relationships may confuse customers and reduce perceived value, negatively impacting purchasing decisions.

#### Chart - 9

````python
# Chart - 9 visualization code
# Chart - 10: Correlation Heatmap

# Goal: Understand relationships between numerical variables

plt.figure(figsize=(8,6))

corr = df[['price', 'MRP', 'discount_percent', 'ratings', 'number_of_ratings']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")

plt.title("Correlation Heatmap of Numerical Features")

plt.show()
````

##### 1. Why did you pick the specific chart?

I chose a heatmap because it provides a clear visual representation of correlations between numerical variables, making it easy to identify strong and weak relationships.

##### 2. What is/are the insight(s) found from the chart?

Price and MRP show a very strong positive correlation (~0.94), indicating they move together. Most other variables, such as ratings and discount_percent, show very weak or near-zero correlation with price, suggesting they are largely independent.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

The strong correlation between price and MRP confirms consistent pricing structure, which is positive for maintaining pricing transparency. However, weak correlation between price and ratings suggests that higher-priced products do not guarantee better customer satisfaction, highlighting the need to focus on product quality rather than pricing alone.
